```python
nums = [12, 45, 7, 45, 30]

largest = second = None

for n in nums:
    if largest is None or n > largest:
        second = largest
        largest = n
    elif n != largest and (second is None or n > second):
        second = n

print(second)
```

```python
a = [1, 4, 7]
b = [2, 3, 8, 9]

i = 0
j = 0
result = []

while i < len(a) and j < len(b):
    if a[i] < b[j]:
        result.append(a[i])
        i += 1
    else:
        result.append(b[j])
        j += 1

result.extend(a[i:])
result.extend(b[j:])

print(result)
```

```python
marks = [
    [78, 85, 62, 90],
    [55, 70, 68, 74],
    [92, 88, 95, 81]
]

for i in range(len(marks)):
    total = sum(marks[i])
    average = total / len(marks[i])

    print(f"Student {i+1}: total = {total}, average = {average:.2f}")
```

```python
marks = [
    [78, 85, 62, 90],
    [55, 70, 68, 74],
    [92, 88, 95, 81]
]

for j in range(4):
    total = 0

    for i in range(3):
        total += marks[i][j]

    average = total / 3
    print(f"Subject {j+1}: {average:.2f}")
```
