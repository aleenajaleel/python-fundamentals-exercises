```python
#A class has 47 students. Each group contains 6 students. Find complete groups and students left over. (Use // and %)
```

```python
total_students=47
students_in_group=6
print("complete groups are",total_students//students_in_group)
print("students left over are",total_students%students_in_group)
```

```python
#Accept a three-digit number and extract hundreds, tens, and units digits using arithmetic operators
```
