```python
#18.	Electricity Usage Alert
#Input units consumed.
#	Units > 500 → "Very High Usage"
#	Units between 300 and 500 → "High Usage"
#	Units between 100 and 299 → "Moderate Usage"
#	Otherwise → "Low Usage"
#Add another condition:
#	If units > 500 or units < 0 → "Invalid/Check Meter Reading"
```

```python
units_consumed=int(input("enter units consumed: "))
if units_consumed>500:
    print("Very High Usage")
elif units_consumed>=300 and units_consumed<=500:
    print("High Usage")
elif units_consumed>=100 and units_consumed<=299:
    print("Moderate Usage")
else:
    print("Low Usage")
if units_consumed>500 or units_consumed<0:
    print("Invalid/Check Meter Reading")
```
